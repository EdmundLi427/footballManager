```python
import requests
import pandas as pd

season = 2026
url = f"https://sports.core.api.espn.com/v3/sports/football/nfl/seasons/{season}/athletes"

all_athletes = []
page = 1
while True:
    resp = requests.get(url, params={"limit": 1000, "page": page}, timeout=30)
    resp.raise_for_status()
    data = resp.json()
    all_athletes.extend(data["items"])
    if page >= data["pageCount"]:
        break
    page += 1

df = pd.json_normalize(all_athletes)
df.to_csv("athletes.csv", index=False)
print(f"Rows: {len(df)}")
```
